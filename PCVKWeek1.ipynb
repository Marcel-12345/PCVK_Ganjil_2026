```python
# langkah 1 ---

import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow # for image display
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt

# langkah 2 ---

# Membuat list untuk menyimpan url dari beberapa image
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
"https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
"https://static.wikia.nocookie.net/gensin-impact/images/8/88/Character_Lynette_Game.png/revision/latest/scale-to-width/360?cb=20241004005040"]
# baca dan tampilkan image
# loop pada tiap url image, beberapa image dapat disimpan pada list
for url in urls:
  image = io.imread(url)
#read image
  image = cv.resize(image, (0,0), fx = 0.5, fy=0.5)
#resize image to half size
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
#convert color to RGB
  final_frame = cv.hconcat((image, image_2))
#concatenate image
  cv2_imshow(final_frame)
#show image
print('\n')

# langkah 3 ---

tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ",tinggi," x ",lebar)
cv2_imshow(image_2)

# langkah 4 ---

image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
#membuat garis horizontal ditengah image
for y in range (lebar):
  image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)

# tugas
print("Tugas praktikum")

# 1. Vertikal
print("Vertikal")

image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# Membuat garis vertikal di tengah image
for x in range(tinggi):
    image_3[x, int(lebar / 2)] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)

# 2. X
print("X")

image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# Membuat garis X (diagonal utama dan sekunder)
for i in range(min(tinggi, lebar)):
    # Garis dari kiri-atas ke kanan-bawah
    image_3[int(i * tinggi / min(tinggi, lebar)), int(i * lebar / min(tinggi, lebar))] = [255, 255, 255]
    # Garis dari kiri-bawah ke kanan-atas
    image_3[int((min(tinggi, lebar) - 1 - i) * tinggi / min(tinggi, lebar)), int(i * lebar / min(tinggi, lebar))] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)

# 3. Garis di tengah
print("Garis di tengah")

image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# Membuat garis horizontal pendek di bagian tengah
panjang_garis = int(lebar * 0.3)  # Panjang garis misal 30% dari lebar gambar
start_x = int((lebar - panjang_garis) / 2)
end_x = start_x + panjang_garis
posisi_y = int(tinggi / 2)

for y in range(start_x, end_x):
    image_3[posisi_y, y] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)

# 4. Persegi di tengah
print("Persegi di tengah")

image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# Ukuran persegi (misal 100x100 pixel)
ukuran = 100

start_y = int((tinggi - ukuran) / 2)
end_y = start_y + ukuran
start_x = int((lebar - ukuran) / 2)
end_x = start_x + ukuran

# Mengisi area persegi di tengah dengan warna putih
for y in range(start_y, end_y):
    for x in range(start_x, end_x):
        image_3[y, x] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

3. Pada uji coba langkah ke-9 terdapat potongan kode program sebagai berikut :
> *image = cv.resize(image, (0,0), fx = 0.5, fy=0.5)*

Apa kegunaan kode program tersebut? dan apa pengaruhnya jika tidak
dilakukan?

Kegunaan: Kode ini digunakan untuk mengubah ukuran (dimensi) gambar menjadi 50% dari ukuran aslinya (setengah kali lebih kecil) secara proporsional.

(0,0) menandakan bahwa kita tidak menentukan dimensi pixel (lebar, tinggi) secara manual.

fx = 0.5 memperkecil lebar gambar sebesar 50%.

fy = 0.5 memperkecil tinggi gambar sebesar 50%.

Pengaruh jika tidak dilakukan:

Gambar akan diproses dengan ukuran dimensi aslinya.

Jika gambar awal berukuran sangat besar, proses manipulasi citra/komputasi akan terasa lebih lambat dan memakan lebih banyak memori RAM.

Tampilan gambar saat ditayangkan di layar mungkin terlalu besar hingga melebihi batas resolusi monitor.

4. Perhatikan potongan kode progam berikut :
> *for y in range (lebar):*
  *image_3[int((tinggi)/2),y] = [255,255,255]*

Apakah kegunaan kode [255,255,255] ? Jelaskan!

Kegunaan: Kode [255,255,255] digunakan untuk mengubah warna pixel di posisi tertentu (image_3[int((tinggi)/2),y]) menjadi warna putih.

Penjelasan: Dalam perpustakaan OpenCV (Python), format warna standar yang digunakan adalah BGR (Blue, Green, Red) dengan rentang nilai intensitas tiap kanal mulai dari 0 hingga 255. Nilai [255, 255, 255] berarti:

Intensitas Blue (Biru) = 255 (maksimal)

Intensitas Green (Hijau) = 255 (maksimal)

Intensitas Red (Merah) = 255 (maksimal)

Pengabungan dari ketiga warna dasar dengan intensitas maksimum ini menghasilkan warna putih solid. Karena kode tersebut dijalankan dalam perulangan for y in range(lebar), potongan kode ini berfungsi untuk menggambar satu garis horizontal berwarna putih tepat di tengah-tengah tinggi gambar.

5. Jelaskan keterkaitan antara pixel dan juga resolusi gambar yang tinggi ataupun rendah!

Pixel (picture element) adalah satuan titik terkecil pembentuk citra digital yang menyimpan nilai warna. Hubungan antara pixel dan resolusi adalah sebagai berikut:Kerapatan dan Jumlah Pixel: Resolusi mendefinisikan seberapa banyak pixel yang terkandung dalam suatu gambar (dimensi lebar * tinggi). Resolusi Tinggi: Memiliki jumlah pixel yang sangat banyak (kerapatan pixel tinggi) dalam satu satuan area. Efeknya, detail gambar terlihat sangat tajam, halus, dan tidak pecah saat diperbesar (zoom). Resolusi Rendah: Memiliki jumlah pixel yang relatif sedikit. Gambar dengan resolusi rendah terlihat kurang detail, kasar, dan jika diperbesar akan tampak kotak-kotak (pixelated) karena setiap pixel-nya terdistribusi pada area yang lebih luas.
